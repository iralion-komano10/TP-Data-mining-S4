```python
# Import des packages
import pandas as pd
```

```python
# Charger le dataset Titanic
df = pd.read_csv("titanic.csv")
```

```python
# Afficher les 10 premières lignes
df.head(10)
```

```python
# Afficher la taille du dataset
df.shape
```

```python
# Informations générales
df.info()
```

```python
# Informations générales
df.describe()
```

```python
# Moyenne de survie par classe
df.groupby('Pclass')['Survived'].mean()
```

```python
# Moyenne de survie par sexe
df.groupby('Sex')['Survived'].mean()
```

```python
# Valeurs manquantes
df.isnull().sum()
```

```python
# Remplacer les valeurs manquantes de Age
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Remplacer les valeurs manquantes de Embarked
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
# Vérification finale
df.isnull().sum()
```
